Việc chọn metric nào phụ thuộc vào **mục tiêu** và **hậu quả** của dự đoán sai trong từng bài toán cụ thể:

## **Khi nào ưu tiên RECALL?**

### Tình huống: **Chi phí bỏ lỡ (False Negative) cao hơn chi phí báo động nhầm (False Positive)**

**Ví dụ thực tế:**
- **Chẩn đoán y tế:** Phát hiện ung thư, COVID-19
  - Bỏ lỡ ca bệnh (FN) → có thể chết người
  - Chẩn đoán nhầm (FP) → làm thêm test, lo lắng thừa

- **Phát hiện gian lận:** Thẻ tín dụng, giao dịch ngân hàng
  - Bỏ lỡ gian lận (FN) → mất tiền thật
  - Báo nhầm (FP) → khách hàng hơi phiền

- **Hệ thống bảo mật:** Phát hiện xâm nhập mạng
  - Bỏ lỡ tấn công (FN) → hệ thống bị hack
  - Báo động nhầm (FP) → admin kiểm tra thêm

**Mục tiêu:** "Không được bỏ lỡ bất kỳ trường hợp nào!"

## **Khi nào ưu tiên PRECISION?**

### Tình huống: **Chi phí báo động nhầm (False Positive) cao hơn chi phí bỏ lỡ (False Negative)**

**Ví dụ thực tế:**
- **Email spam:** Lọc email spam
  - Chặn nhầm email quan trọng (FP) → mất cơ hội việc làm, hợp đồng
  - Để spam vào inbox (FN) → chỉ hơi phiền

- **Hệ thống recommend:** Gợi ý sản phẩm, nội dung
  - Gợi ý sai (FP) → khách hàng mất niềm tin
  - Không gợi ý được sản phẩm tốt (FN) → ít ảnh hưởng

- **Quảng cáo targeting:** Hiển thị quảng cáo cho đúng đối tượng
  - Show sai người (FP) → lãng phí budget, phiền người dùng
  - Bỏ lỡ khách tiềm năng (FN) → chấp nhận được

**Mục tiêu:** "Những gì dự đoán phải chính xác cao!"

## **Khi nào sử dụng F1 SCORE?**

### Tình huống: **Cần cân bằng giữa Precision và Recall**

**Ví dụ thực tế:**
- **Tìm kiếm thông tin:** Search engine, IR system
  - Cần tìm được nhiều kết quả liên quan (recall cao)
  - Đồng thời kết quả phải chính xác (precision cao)

- **Phân loại văn bản:** Phân loại tin tức, sentiment analysis
  - Không có lý do đặc biệt để ưu tiên một metric

- **Object detection:** Phát hiện vật thể trong ảnh
  - Cần detect được nhiều object (recall)
  - Những detection phải chính xác (precision)

- **Bài toán imbalanced dataset:** Khi số lượng class không cân bằng
  - F1 score ít bị ảnh hưởng bởi class imbalance hơn accuracy

**Mục tiêu:** "Cần hiệu suất tổng thể tốt, không thiên lệch về một phía"

## **Bảng tóm tắt quyết định:**

| Tình huống | Metric ưu tiên | Lý do |
|------------|---------------|-------|
| Bỏ lỡ = thảm họa | **Recall** | Chi phí FN >> chi phí FP |
| Báo nhầm = thảm họa | **Precision** | Chi phí FP >> chi phí FN |
| Cần cân bằng | **F1 Score** | Chi phí FN ≈ chi phí FP |
| Dataset cân bằng, tổng quan | **Accuracy** | Đơn giản, dễ hiểu |
